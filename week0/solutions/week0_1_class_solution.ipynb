## 파트 A. 클래스 기초

클래스는 **"데이터(속성)"와 "기능(메서드)"을 하나로 묶은 설계도**입니다. 설계도로 만든 실물을 **객체(인스턴스)**라고 합니다.

```python
class Dog:
    def __init__(self, name):   # 객체를 만들 때 자동 실행
        self.name = name        # self.xxx = 객체가 기억하는 값

    def bark(self):             # 메서드: 첫 번째 인자는 항상 self
        return self.name + ": 멍!"

d = Dog("초코")    # __init__(self=d, name="초코") 가 실행됨
d.bark()          # '초코: 멍!'
```

- `self`는 **"지금 이 객체 자신"**을 가리킵니다.
- `__init__`은 **생성자**입니다. 객체가 처음 만들어질 때 한 번 실행됩니다.

````python
# 예제를 직접 실행해 보세요
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return self.name + ": 멍!"

d1 = Dog("초코")
d2 = Dog("콩이")
print(d1.bark())
print(d2.bark())
print(d1.name, d2.name)   # 객체마다 자기 값을 따로 기억합니다
````

### 문제 A-1. 카운터 만들기

`Counter` 클래스를 완성하세요.
- 만들어질 때 `count`가 0에서 시작합니다.
- `increase()`를 부르면 `count`가 1 늘어납니다.
- `get()`은 현재 `count`를 돌려줍니다.

````python
class Counter:
    def __init__(self):
        self.count = 0

    def increase(self):
        self.count += 1

    def get(self):
        return self.count
````

````python
def test():
    c = Counter()
    assert c.get() == 0, "처음에는 0이어야 해요"
    c.increase(); c.increase()
    assert c.get() == 2, "increase()를 두 번 부르면 2여야 해요"
    c2 = Counter()
    assert c2.get() == 0, "새 객체는 다른 객체와 상관없이 0에서 시작해야 해요"

check(test)
````

### 문제 A-2. 학생 성적

`Student` 클래스를 완성하세요.
- `Student("세홍", [90, 80, 70])`처럼 이름과 점수 리스트를 받아 저장합니다.
- `average()`는 평균 점수를 돌려줍니다.
- `add_score(score)`는 점수를 하나 추가합니다.

````python
class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def average(self):
        return sum(self.scores) / len(self.scores)

    def add_score(self, score):
        self.scores.append(score)
````

````python
def test():
    s = Student("세홍", [90, 80, 70])
    assert s.name == "세홍", "이름이 저장되지 않았어요"
    assert s.average() == 80, "평균이 80이어야 해요"
    s.add_score(100)
    assert s.average() == 85, "100점을 추가하면 평균이 85여야 해요"

check(test)
````

### 문제 A-3. 은행 계좌 (조건이 있는 메서드)

`BankAccount`를 완성하세요.
- `BankAccount(owner, balance=0)`: 잔액 기본값은 0
- `deposit(amount)`: 잔액에 더하기
- `withdraw(amount)`: 잔액이 충분하면 빼고 `True`를, 부족하면 **빼지 않고** `False`를 돌려주기

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            return False
        self.balance -= amount
        return True
````

````python
def test():
    a = BankAccount("세홍")
    assert a.balance == 0, "기본 잔액은 0이어야 해요"
    a.deposit(1000)
    assert a.withdraw(300) is True and a.balance == 700, "300을 빼면 True, 잔액 700이어야 해요"
    assert a.withdraw(5000) is False and a.balance == 700, "잔액이 부족하면 False, 잔액은 그대로여야 해요"

check(test)
````

## 파트 B. 상속과 `super()`

**상속**은 이미 있는 클래스(부모)의 기능을 물려받아 새 클래스(자식)를 만드는 것입니다.

```python
class Animal:                 # 부모
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(Animal):            # 자식: 괄호 안에 부모 이름
    def speak(self):          # 오버라이딩: 부모 메서드를 덮어쓰기
        return "멍멍"
```

`Dog`에는 `__init__`을 안 썼지만 부모 것을 물려받아서 `Dog("초코").name`이 동작합니다.

**PyTorch에서 모델을 만드는 것 = `nn.Module`을 상속받는 것**입니다. 그래서 상속이 중요합니다.

### 문제 B-1. 오버라이딩

`Animal`을 상속받아 `Dog`와 `Cat`을 만드세요.
- `Dog().speak()`는 `"멍멍"`, `Cat().speak()`는 `"야옹"`
- `Animal`의 `introduce()`는 그대로 물려받아서 `"초코: 멍멍"`처럼 나와야 합니다. 직접 다시 쓰지 마세요.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def introduce(self):
        return self.name + ": " + self.speak()


class Dog(Animal):
    def speak(self):
        return "멍멍"


class Cat(Animal):
    def speak(self):
        return "야옹"
````

````python
def test():
    assert issubclass(Dog, Animal) and issubclass(Cat, Animal), "Dog, Cat 은 Animal 을 상속받아야 해요"
    assert Dog("초코").speak() == "멍멍" and Cat("나비").speak() == "야옹", "speak() 결과가 달라요"
    assert Dog("초코").introduce() == "초코: 멍멍", "introduce() 는 부모 것을 그대로 쓰면 돼요"
    assert "introduce" not in Dog.__dict__, "introduce 를 Dog 안에 다시 쓰지 말고 물려받으세요"

check(test)
````

### 문제 B-2. `super().__init__()` ★ 가장 중요

자식이 **자기만의 `__init__`**을 가지면 부모의 `__init__`은 자동으로 실행되지 **않습니다**.
그래서 `super().__init__(...)`으로 부모의 준비 작업을 직접 불러줘야 합니다.

`Dog`를 완성하세요.
- `Dog(name, breed)`: `name`은 **부모(`Animal`)에게 맡기고**, `breed`(품종)만 직접 저장합니다.
- 반드시 `super().__init__(name)`을 사용하세요.

````python
class Animal:
    def __init__(self, name):
        self.name = name
        self.alive = True


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
````

````python
def test():
    d = Dog("초코", "푸들")
    assert getattr(d, "breed", None) == "푸들", "breed 가 저장되지 않았어요"
    assert getattr(d, "name", None) == "초코", "name 이 없어요. super().__init__(name) 을 불렀나요?"
    assert getattr(d, "alive", None) is True, "alive 가 없어요. 부모의 준비 작업이 실행되지 않았어요"

check(test)
````

## 파트 C. 클래스로 PyTorch 맛보기

### 특수 메서드 `__call__`

PyTorch에서는 `model(x)`처럼 **객체를 함수처럼 부릅니다.** 이게 가능한 이유는 `__call__` 때문입니다.

```python
class Greeter:
    def __call__(self, name):
        return "안녕, " + name

g = Greeter()
g("세홍")     # 내부적으로 g.__call__("세홍") 이 실행됨
```

`nn.Module`은 `__call__` 안에서 우리가 만든 `forward()`를 불러줍니다. 그래서 우리는 `forward`만 작성하면 됩니다.

### 문제 C-1. 나만의 미니 `nn.Module` 흉내내기

아래 `MiniModule`은 `nn.Module`을 아주 단순하게 흉내 낸 부모 클래스입니다. `obj(x)`를 부르면 `obj.forward(x)`를 대신 실행합니다.

1. `MiniModule`의 `__call__`을 완성하세요: `self.forward(x)`의 결과를 돌려주면 됩니다.
2. `MiniModule`을 상속받아 `Scale`을 만드세요.
   - `Scale(k)`: 숫자 `k`를 저장 (부모 `__init__`도 불러주기)
   - `forward(x)`: `x * k`를 돌려주기
3. 결과적으로 `Scale(3)(5)`가 `15`가 되어야 합니다.

````python
class MiniModule:
    def __init__(self):
        self.call_count = 0

    def __call__(self, x):
        self.call_count += 1
        return self.forward(x)

    def forward(self, x):
        raise NotImplementedError("자식 클래스에서 forward 를 만들어야 해요")


class Scale(MiniModule):
    def __init__(self, k):
        super().__init__()
        self.k = k

    def forward(self, x):
        return x * self.k
````

````python
def test():
    s = Scale(3)
    assert isinstance(s, MiniModule), "Scale 은 MiniModule 을 상속받아야 해요"
    assert s(5) == 15, "Scale(3)(5) 는 15여야 해요"
    assert s.call_count == 1, "call_count 가 1이어야 해요. super().__init__() 을 불렀나요?"

check(test)
````

### 문제 C-2. 진짜 `nn.Module`로 MNIST용 모델 만들기

이제 진짜 PyTorch입니다. 구조는 C-1과 똑같습니다.

`MLP`를 완성하세요.
- `__init__`: `self.fc1 = nn.Linear(784, 256)`, `self.relu = nn.ReLU()`, `self.fc2 = nn.Linear(256, 10)`
- `forward(x)`: `fc1 → relu → fc2` 순서로 통과시키기

28×28 손글씨 이미지(784픽셀)를 받아 숫자 10개 중 하나로 분류하는 모델입니다. 시즌2 Lab 09-2와 같은 구조입니다.

````python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 256)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(256, 10)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        return self.fc2(x)
````

````python
def test():
    model = MLP()
    x = torch.randn(32, 784)          # 이미지 32장 (배치 크기 32)
    out = model(x)
    assert out.shape == (32, 10), f"출력 shape 이 (32, 10) 이어야 하는데 {tuple(out.shape)} 이에요"
    assert isinstance(model.fc1, nn.Linear) and model.fc1.out_features == 256, "fc1 은 Linear(784, 256) 이어야 해요"

check(test)
````

### 문제 C-3. 파라미터 개수 세기

`nn.Module`을 상속받으면 `model.parameters()`로 **학습할 숫자(가중치, 편향)**를 전부 꺼낼 수 있습니다. 부모가 `__init__`에서 이걸 추적하도록 준비해 주기 때문입니다.

1. 코드로 세기: `sum(p.numel() for p in model.parameters())`
2. **손으로 계산**해서 `my_answer`에 적기. `Linear(a, b)`의 파라미터 개수는 가중치 `a*b`개와 편향 `b`개입니다.

````python
model = MLP()

n_params = sum(p.numel() for p in model.parameters())
my_answer = (784 * 256 + 256) + (256 * 10 + 10)   # fc1 + fc2 (ReLU 는 파라미터가 없음)

print(n_params, my_answer)
````

````python
def test():
    assert n_params == 203530, f"코드 계산 결과가 203530 이어야 해요 (지금: {n_params})"
    assert my_answer == 203530, "손 계산이 틀렸어요. 가중치 a*b, 편향 b 를 층마다 더하세요"

check(test)
````

### 문제 C-4. ResNet 미리보기: 잔차 연결(skip connection)

ResNet의 핵심 아이디어는 **출력에 입력을 그대로 더해주는 것**입니다.

$$\text{out} = F(x) + x$$

`ResidualBlock`을 완성하세요.
- `__init__(self, dim)`: `self.fc = nn.Linear(dim, dim)`, `self.relu = nn.ReLU()`
- `forward(x)`: `self.relu(self.fc(x)) + x`

채점 셀에서는 `fc`의 가중치를 모두 0으로 만들었을 때 **출력이 입력과 똑같은지** 확인합니다. 층이 아무것도 배우지 못해도 최소한 입력은 그대로 전달된다는 것, 이게 ResNet이 깊어져도 학습이 잘 되는 이유입니다.

````python
class ResidualBlock(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.fc = nn.Linear(dim, dim)
        self.relu = nn.ReLU()

    def forward(self, x):
        return self.relu(self.fc(x)) + x
````

````python
def test():
    block = ResidualBlock(8)
    x = torch.randn(4, 8)
    assert block(x).shape == (4, 8), "입력과 출력의 shape 이 같아야 해요"
    with torch.no_grad():
        block.fc.weight.zero_()
        block.fc.bias.zero_()
    assert torch.equal(block(x), x), "가중치가 0이면 출력이 입력과 같아야 해요. + x 를 했나요?"

check(test)
````

### 문제 C-5. Stanford40 미리보기: 나만의 `Dataset`

나중에 Stanford40 데이터를 불러올 때 `torch.utils.data.Dataset`을 상속받은 클래스를 만들게 됩니다. 꼭 필요한 특수 메서드는 두 개입니다.
- `__len__(self)`: 데이터 개수 → `len(dataset)`이 동작함
- `__getitem__(self, idx)`: `idx`번째 (입력, 정답) → `dataset[idx]`가 동작함

`SquareDataset(n)`을 완성하세요.
- 데이터는 0부터 `n-1`까지의 정수입니다.
- `dataset[i]`는 `(torch.tensor([float(i)]), torch.tensor([float(i**2)]))`를 돌려줍니다. 입력은 i, 정답은 i의 제곱입니다.

````python
from torch.utils.data import Dataset, DataLoader

class SquareDataset(Dataset):
    def __init__(self, n):
        self.n = n

    def __len__(self):
        return self.n

    def __getitem__(self, idx):
        x = torch.tensor([float(idx)])
        y = torch.tensor([float(idx ** 2)])
        return x, y
````

````python
def test():
    ds = SquareDataset(10)
    assert len(ds) == 10, "len(ds) 가 10이어야 해요"
    x, y = ds[3]
    assert x.item() == 3.0 and y.item() == 9.0, "ds[3] 은 (3, 9) 여야 해요"

    # DataLoader 는 Dataset 에서 여러 개를 꺼내 배치로 묶어줍니다
    loader = DataLoader(ds, batch_size=4, shuffle=False)
    xb, yb = next(iter(loader))
    assert xb.shape == (4, 1), f"배치 shape 이 (4, 1) 이어야 해요 (지금: {tuple(xb.shape)})"
    print("첫 배치 입력:", xb.view(-1).tolist(), " 정답:", yb.view(-1).tolist())

check(test)
````
