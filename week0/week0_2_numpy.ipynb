## 파트 C. 연산, 브로드캐스팅, axis

### 원소별 곱 `*` vs 행렬곱 `@`
```python
A = np.array([[1, 2], [3, 4]])
A * A    # [[1, 4], [9, 16]]   → 같은 자리끼리 곱
A @ A    # [[7, 10], [15, 22]] → 행렬곱
```
행렬곱 `(n, k) @ (k, m)`의 결과는 `(n, m)`입니다. **가운데 k가 같아야** 합니다.

### axis
```python
x = np.array([[1, 2, 3],
              [4, 5, 6]])
x.sum(axis=0)   # [5, 7, 9]  → 위아래로 합침 (0번 축이 사라짐) → shape (3,)
x.sum(axis=1)   # [6, 15]    → 좌우로 합침 (1번 축이 사라짐) → shape (2,)
```
요령: **`axis=k`라고 하면 shape에서 k번째 숫자가 사라진다.**

## 파트 D. NumPy에서 PyTorch로

PyTorch 텐서는 NumPy와 거의 같습니다. 이름이 조금 다른 것만 기억하면 됩니다.

| NumPy | PyTorch |
|---|---|
| `np.array(...)` | `torch.tensor(...)` |
| `a.shape` | `t.shape` 또는 `t.size()` |
| `a.reshape(2, -1)` | `t.view(2, -1)` 또는 `t.reshape(2, -1)` |
| `a.sum(axis=0)` | `t.sum(dim=0)` |
| `a.argmax(axis=1)` | `t.argmax(dim=1)` |
| `a @ b` | `t @ u` 또는 `t.matmul(u)` |
| 서로 변환 | `torch.from_numpy(a)`, `t.numpy()` |

**PyTorch만 가진 기능:** GPU 계산(`.to('cuda')`)과 **자동 미분**(`requires_grad=True`, `.backward()`).

2. `cost.backward()`를 부르면 `W.grad`에 **cost를 줄이려면 W를 어느 방향으로 바꿔야 하는지**(기울기)가 계산됩니다.
